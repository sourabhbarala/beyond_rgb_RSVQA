```python
def fill_numeric_ans(
    parent_dir_path,
    sensor,
    # num_ques
    ):
    
    
    def delete_json(
        path:str
    ):
        import os
        
        if path.endswith("generated_answers_update_num.json"):
            if os.path.exists(path):
                os.remove(path)
                print(f"Deleted file: {path}")
            else:
                print(f"File not found: {path}")
            
            
    
    def save_in_json(
            content:dict,
            path:str,
            file_name:str
            ):
            """Saves a dictionary as json file.

            Args:
                content (dict): Dictionary containing the content
                path (str): The directory in which json file will be saved, for eg. "C/abc/.../sentinel/all_bands"
                file_name (str): Name of the json file, for eg. "water_info.json"
            """
            
            with open(os.path.join(path,file_name), "w") as f:
                json.dump(content, f)
                
    cell_map={
            'cell_1':"upper left",
            'cell_2':"upper middle",
            'cell_3':"upper right",
            'cell_4':"middle left",
            'cell_5':"center",
            'cell_6':"middle right",
            'cell_7':"lower left",
            'cell_8':"lower middle",
            'cell_9':"lower right"}
    
    bands_dir_names=os.listdir(parent_dir_path)
    
    for i,band_dir_name in enumerate(bands_dir_names):
    
        print(i," : ",band_dir_name)
        bands_dir_path=os.path.join(
            parent_dir_path,
            band_dir_name
        )
        
        
        
        # * open answers json file
        ans_json_data=None
        ans_file_name="generated_answers.json"
        ans_file_path=os.path.join(
            bands_dir_path,
            ans_file_name
        )
        
        with open(ans_file_path, "r", encoding="utf-8") as f:
                
            ans_json_data=json.load(f)
        
        # ===========================================================================================================
        
        # * open veg_stats
        veg_stats_data=None
        veg_stats_name="veg_stats.json"
        veg_stats_path=os.path.join(
            bands_dir_path,
            veg_stats_name
        )
        
        with open(veg_stats_path, "r", encoding="utf-8") as f:
                
            veg_stats_data=json.load(f)
            
        # ===========================================================================================================
        
        
        # * open built_up_stats
        built_up_stats_data=None
        if sensor!='lis4':
            
            built_up_stats_name="built_up_stats.json"
            built_up_stats_path=os.path.join(
                bands_dir_path,
                built_up_stats_name
            )
            
            with open(built_up_stats_path, "r", encoding="utf-8") as f:
                    
                built_up_stats_data=json.load(f)
            
        # ===========================================================================================================
        
        # * open water_info
        water_info_data=None
        water_info_name="water_bodies_info.json"
        water_info_path=os.path.join(
            bands_dir_path,
            water_info_name
        )
        
        with open(water_info_path, "r", encoding="utf-8") as f:
                
            water_info_data=json.load(f)
            
        # ===========================================================================================================
        
        # * open overall_land_stats
        overall_land_stats_data=None
        overall_land_stats_name="overall_land_stats.json"
        overall_land_stats_path=os.path.join(
            bands_dir_path,
            overall_land_stats_name
        )
        
        with open(overall_land_stats_path, "r", encoding="utf-8") as f:
                
            overall_land_stats_data=json.load(f)
        
        
        # ===========================================================================================================
        # ===========================================================================================================
        
        updated_ques_ans=ans_json_data.copy()
        
        #> What is % of the area covered by vegetation?
        ans = overall_land_stats_data["over all land cover %"]["info"]["moderate vegetation coverage"] + \
        overall_land_stats_data["over all land cover %"]["info"]["dense vegetation coverage"]
        
        updated_ques_ans["What is % of the area covered by vegetation?".strip()] = ans
        
        del ans
        
        # ============================================================================================================
        
        #> What is % of the area covered by dense vegetation?
        ans = overall_land_stats_data["over all land cover %"]["info"]["dense vegetation coverage"]
        
        updated_ques_ans["What is % of the area covered by dense vegetation?".strip()] = ans
        
        del ans
        
        # ============================================================================================================
        
        #> Which area in the image has the most vegetation (upper left, upper middle, upper right, middle left, center ,middle right, lower left, lower middle, lower right)?
        ans = None
        max_veg_cell=None
        max_veg=0
        
        for cell, cell_data in veg_stats_data.items():
            
            total_veg = 0
            cell_total_area=0
            if sensor!='sentinel':
                cell_total_area = cell_data["area covered by cell"]
                
            else:
                cell_total_area = cell_data["total area covered by cell in km2"]
            
            keys_in_coverage_info=list(cell_data["vegetation class coverage %"]["coverage_info"].keys())
            
            if "moderate vegetation" in keys_in_coverage_info:
                moderate_veg_area=cell_data["vegetation class coverage %"]["coverage_info"]["moderate vegetation"]*(cell_total_area/100)
                
                total_veg += moderate_veg_area
            
            if "dense vegetation" in keys_in_coverage_info:
                dense_veg_area=cell_data["vegetation class coverage %"]["coverage_info"]["dense vegetation"]*(cell_total_area/100)
            
                total_veg += dense_veg_area
            
            if total_veg>max_veg:
                max_veg=total_veg
                max_veg_cell=cell
                
        ans= cell_map[max_veg_cell]
        updated_ques_ans["Which area in the image has the most vegetation (upper left, upper middle, upper right, middle left, center ,middle right, lower left, lower middle, lower right)?".strip()] = ans
        
        del ans
        
        # ============================================================================================================
        
        #> What is the total area covered by water (number + units)?
        total_w_area=0
        
        for wbn, data_dict in water_info_data.items():
            total_w_area += data_dict["area in m2"]
            
        updated_ques_ans["What is the total area covered by water (number + units)?".strip()] = [
            f"{total_w_area} m2",
            f"{total_w_area} m^2",
            f"{total_w_area} square meter", 
            f"{total_w_area/1000000} km2", 
            f"{total_w_area/1000000} km^2", 
            f"{total_w_area/1000000} square kilometers"]
        
        # ============================================================================================================
        
        #> What is the area of the largest water body?
        all_wb_area=[0]
        
        for wbn, data_dict in water_info_data.items():
            all_wb_area.append(data_dict["area in m2"])
            
        updated_ques_ans["What is the area of the largest water body?".strip()] = [
            f"{max(all_wb_area)} m2",
            f"{max(all_wb_area)} square meters",
            f"{max(all_wb_area)} m^2"
            f"{max(all_wb_area)/1000000} km2",
            f"{max(all_wb_area)/1000000} km^2",
            f"{max(all_wb_area)/1000000} square kilometers"]
        
        # ============================================================================================================
        
        #> What % of area is covered by built-up structures?
        if sensor!='lis4':
            ans = overall_land_stats_data["over all land cover %"]["info"]["moderate built-up coverage"] + \
            overall_land_stats_data["over all land cover %"]["info"]["dense built-up coverage"]
            
            updated_ques_ans["What % of area is covered by built-up structures?".strip()] = ans
            
            del ans
        
        # ============================================================================================================
        
        #> Which part of the image has the most built-up area (upper left, upper middle, upper right, middle left, center ,middle right, lower left, lower middle, lower right)?
        if sensor!='lis4':
            ans = None
            max_built_cell=None
            max_built=0
            for cell, cell_data in built_up_stats_data.items():
                
                total_built=0
                cell_total_area=cell_data["total area covered by cell in km2"]
                
                keys_in_coverage_info=list(cell_data["built-up class coverage %"]["coverage"].keys())
                
                if "Medium-density built-up/stabilized desert" in keys_in_coverage_info:
                    moderate_built_area=cell_data["built-up class coverage %"]["coverage"]["Medium-density built-up/stabilized desert"]*(cell_total_area/100)
                    
                    total_built += moderate_built_area
                    
                
                if "High-density built-up areas" in keys_in_coverage_info:
                    dense_built_area=cell_data["built-up class coverage %"]["coverage"]["High-density built-up areas"]*(cell_total_area/100)
                    
                    total_built +=  dense_built_area
                
                if total_built>max_built:
                    max_built=total_built
                    max_built_cell=cell
                    
            ans= cell_map[max_built_cell]
            updated_ques_ans["Which part of the image has the most built-up area (upper left, upper middle, upper right, middle left, center ,middle right, lower left, lower middle, lower right)?".strip()] = ans
            
            del ans
        # ============================================================================================================
    

        # # * delete the file if it already exists
        # delete_json(
        #     path=os.path.join(
        #         bands_dir_path,
        #         "generated_answers_update_num.json"
        #     )
        # )
        # * save the new file
        save_in_json(
            content=updated_ques_ans,
            path=bands_dir_path,
            file_name="generated_answers_update_num.json"
        )
        
    

                
```
